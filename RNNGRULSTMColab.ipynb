```python
import os, shutil

REPO = "/content/emg2qwerty-c147-project"

DRIVE = ("/content/drive/MyDrive/Winter 26 Quarter/"
         "ECE C147 Neural Networks & Deep Learning/147 Final Project")

# Copy ckpt locally to avoid path issues
os.makedirs("/content/ckpts", exist_ok=True)
shutil.copy2(f"{DRIVE}/models/generic.ckpt", "/content/ckpts/generic.ckpt")

# Copy LM files where the decoder expects them
os.makedirs(f"{REPO}/models/lm", exist_ok=True)
for f in ["wikitext-103-6gram-charlm.bin", "wikitext-103-6gram-charlm.arpa"]:
    shutil.copy2(f"{DRIVE}/models/lm/{f}", f"{REPO}/models/lm/{f}")


# Link the data folder into the repo
data_dest = "/content/emg2qwerty-c147-project/data"
if not os.path.exists(data_dest):
    os.symlink(f"{DRIVE}/data", data_dest)

print("data linked:", os.listdir(data_dest)[:3])

print("done, generic size:", os.path.getsize("/content/ckpts/generic.ckpt"))
```

```python
import subprocess

REPO    = "/content/emg2qwerty-c147-project"
GENERIC = "/content/ckpts/generic.ckpt"
logs    = {}


DECODERS = {
    "no_decoder": "ctc_greedy",
    "6gram":      "ctc_beam",
}

for dec_key, dec_cfg in DECODERS.items():
    print(f"\n=== generic | {dec_key} ===")
    result = subprocess.run([
        "python", "-m", "emg2qwerty.train",
        "user=single_user",
        "model=tds_conv_ctc",
        "train=False",
        "trainer.accelerator=gpu",
        "trainer.devices=1",
        f"decoder={dec_cfg}",
        f"checkpoint={GENERIC}",
    ], cwd=REPO, capture_output=True, text=True)
    combined = result.stdout + result.stderr
    logs[dec_key] = combined
    print("\n".join(combined.strip().splitlines()[-20:]))
```

```python
import subprocess

result = subprocess.run([
    "python", "-m", "emg2qwerty.train",
    "user=single_user",
    "model=tds_conv_ctc",
    "train=False",
    "trainer.accelerator=gpu",
    "trainer.devices=1",
    "decoder=ctc_beam_transformer",
    "checkpoint=/content/ckpts/generic.ckpt",
], cwd="/content/emg2qwerty-c147-project", capture_output=True, text=True)

combined = result.stdout + result.stderr
logs["transformer"] = combined
print("\n".join(combined.strip().splitlines()[-20:]))
```

```python
import re

print(f"{'model':<12} {'decoder':<14} {'CER':>8}")
print("-" * 38)
for dec_key, log in logs.items():
    m = re.search(r"'test/CER':\s*([\d.]+)", log)
    cer = float(m.group(1)) if m else None
    print(f"{'generic':<12} {dec_key:<14} {f'{cer:.4f}' if cer else 'N/A':>8}")
```

```python
import matplotlib.pyplot as plt
import numpy as np

decoders = ['Greedy\n(no decoder)', '6-gram\nBeam Search', 'Transformer LM\nBeam Search']
cer_vals = [7.7156, 3.6093, 7.6724]
colors = ['steelblue', 'green', 'coral']

fig, ax = plt.subplots(figsize=(7, 4))
bars = ax.bar(decoders, cer_vals, color=colors, width=0.5)
ax.set_ylabel('Test CER (%)', fontsize=12)
ax.set_title('CTC Decoder Comparison on Generic TDS Model', fontsize=13)
ax.set_ylim(0, 10)
for bar, val in zip(bars, cer_vals):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 0.1,
            f'{val:.2f}%', ha='center', fontsize=11, fontweight='bold')
plt.tight_layout()
plt.savefig('decoder_comparison.png', dpi=150)
plt.show()
```

```python
import subprocess, re

FINETUNED = "/content/ckpts/finetuned.ckpt"
REPO = "/content/emg2qwerty-c147-project"

DECODERS = {
    "no_decoder":  "ctc_greedy",
    "6gram":       "ctc_beam",
    "transformer": "ctc_beam_transformer",
}

for dec_key, dec_cfg in DECODERS.items():
    print(f"\n=== finetuned | {dec_key} ===")
    result = subprocess.run([
        "python", "-m", "emg2qwerty.train",
        "user=single_user",
        "model=tds_conv_ctc",
        "train=False",
        "trainer.accelerator=gpu",
        "trainer.devices=1",
        f"decoder={dec_cfg}",
        f"checkpoint={FINETUNED}",
    ], cwd=REPO, capture_output=True, text=True)
    combined = result.stdout + result.stderr
    logs[f"finetuned_{dec_key}"] = combined
    m = re.search(r"'test/CER':\s*([\d.]+)", combined)
    print("CER:", m.group(1) if m else "error — check output")
    if not m:
        print("\n".join(combined.strip().splitlines()[-10:]))
```
